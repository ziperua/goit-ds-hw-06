```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression
```

```python
df = pd.read_csv('Housing.csv')
X = df[['area', 'bathrooms', 'bedrooms']]
y = df['price']
```

```python
X = (X-X.mean()) / X.std()
X = np.c_[np.ones(len(X)), X]
```

```python
def hypothesis(X, w):
    return X @ w
```

```python
def loss(X, y, w):
    predictions = hypothesis(X, w)
    errors = predictions - y
    return (errors**2).sum() / (2 * len(y)) 
```

```python
def gradient_step(X, y, w, alpha):
    errors = hypothesis(X, w) - y
    gradient = X.T @ errors / len(y)
    return w - alpha * gradient
```

```python
w = np.zeros(4)
for i in range(1000):
    w = gradient_step(X,y,w,0.1)
```

```python
loss(X,y,w)
```

```python
w_analytic = np.linalg.inv(X.T @ X) @ X.T @ y
w_analytic
```

```python
model = LinearRegression().fit(X[:,1:], y)
print(model.intercept_)
print(model.coef_)
```

```python
np.allclose(w, w_analytic)                                  # спуск vs формула
np.allclose(w[1:], model.coef_)                             # спуск vs sklearn
np.allclose(hypothesis(X, w), model.predict(X[:, 1:]))      # одинаковые ли
```
